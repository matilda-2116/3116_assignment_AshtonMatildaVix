Now I want to go through each measurement that indicates accretion, highlight each globular cluster as either in situ, mixed, or ex situ, combine all globular clusters ranked in terms of which have the most ex situ values/mixed values and then colour those specific GCs on all the graphs as different colours, those for most likely to be accreted, those possibly accreted, and those probably not accreted. This will provide a list of potential accreted globular clusters and will show where they are on all the graphs.

```python
AccretedWeighting = {} # Creates a variable that will contain the accreted weighting sum for each individual 
                        # globular cluster id, which is why we use curly brackets so we can assign the sum to a key, the GCIdentification
```

```python
# AGES

MergedDataArray = np.array(MergedData, dtype=object)
YoungGCAges = [] #list of 0s to add metallicities of low metallicity GCs after they are identified by the loop
YoungGCIndexes = [] #same as above but to add the indexes 
MiddleGCAges = []
MiddleGCIndexes = []
OldGCAges = []
OldGCIndexes = [] #to add other GCs to to help with plotting

GCAges1 = MergedDataArray[:,28]
GCAges2 = MergedDataArray[:,32]

for index, item in enumerate(GCAges1): # going through each GC entry for the entire age data set
    GCIdentification = MergedDataArray[index,0] # assigning the respective globular cluster name to its value, essentially, for the first GC, assign the first ID, for second GC assign second ID etc
    AccretedWeighting[GCIdentification] = 0 # Sets the base sum to 0
    if item < 11: # check parameter value
        AccretedWeighting[GCIdentification] += 5 # if in range, give weighting
        # print out the classification text below
        print('Name =', GCIdentification, ' - Category = Young', ' - Age(Gyr) =', item, ' - Index =', index + 1, ' - Weighting = ', AccretedWeighting[GCIdentification])
        YoungGCAges.append(item) # add age to young GC correct ages set
        YoungGCIndexes.append(index) # add GC index to correct ages set
    elif item < 12:
        AccretedWeighting[GCIdentification] += 3
        print('Name =', GCIdentification, ' - Category = Middle', ' - Age(Gyr) =', item, ' - Index =', index + 1, ' - Weighting = ', AccretedWeighting[GCIdentification])
        YoungGCAges.append(item)
        YoungGCIndexes.append(index)
    else:
        AccretedWeighting[GCIdentification] += 0
        print('Name =', GCIdentification, ' - Category = Old', ' - Age(Gyr) =', item, ' - Index =', index + 1, ' - Weighting = ', AccretedWeighting[GCIdentification])
        OldGCAges.append(item)
        OldGCIndexes.append(index)
        # classifies, weighs, and prints details for every GC in GCAges1, exact same method used for other relevant parameters
```

```python
# GALACTOCENTRIC DISTANCE
MergedDataArray = np.array(MergedData, dtype=object)
CloseGCDistances = [] #list of 0s to add metallicities of low metallicity GCs after they are identified by the loop
CloseGCIndexes = [] #same as above but to add the indexes 
MixedDistanceGCs = []
MixedDistanceGCIndexes = []
FarGCDistances = []
FarGCIndexes = [] #to add other GCs to to help with plotting

GCGalactocentricDistance = MergedDataArray[:,7]

for index, item in enumerate(GCGalactocentricDistance):
    GCIdentification = MergedDataArray[index,0]
    if item > 11:
        AccretedWeighting[GCIdentification] += 4
        print('Name =', GCIdentification, ' - Category = Far', ' - Distance(kpc) =', item, ' - Index =', index + 1, ' - Weighting = ', AccretedWeighting[GCIdentification])
        FarGCDistances.append(item)
        FarGCIndexes.append(index)
    elif item > 8:
        AccretedWeighting[GCIdentification] += 2
        print('Name =', GCIdentification, ' - Category = Middle', ' - Distance(kpc) =', item, ' - Index =', index + 1, ' - Weighting = ', AccretedWeighting[GCIdentification])
        MixedDistanceGCs.append(item)
        MixedDistanceGCIndexes.append(index)
    else:
        AccretedWeighting[GCIdentification] += 0
        print('Name =', GCIdentification, ' - Category = Close', ' - Distance(kpc) =', item, ' - Index =', index + 1, ' - Weighting = ', AccretedWeighting[GCIdentification])
        CloseGCDistances.append(item)
        CloseGCIndexes.append(index)
```

```python
# Z DISTANCE
MergedDataArray = np.array(MergedData, dtype=object)
DiskGCHeights = [] #list of 0s to add metallicities of low metallicity GCs after they are identified by the loop
DiskGCIndexes = [] #same as above but to add the indexes 
MixedHeightGCs = []
MixedHeightGCIndexes = []
HighGCHeights = []
HighGCIndexes = [] #to add other GCs to to help with plotting

GCHeights = MergedDataArray[:,10]

for index, item in enumerate(GCHeights):
    GCIdentification = MergedDataArray[index,0]
    if np.abs(item) < 3:
        AccretedWeighting[GCIdentification] += 0
        print('Name =', GCIdentification, ' - Category = Disk', ' - Height(kpc) =', item, ' - Index =', index + 1, ' - Weighting = ', AccretedWeighting[GCIdentification])
        DiskGCHeights.append(item)
        DiskGCIndexes.append(index)
    elif np.abs(item) < 6:
        AccretedWeighting[GCIdentification] += 2
        print('Name =', GCIdentification, ' - Category = Mixed', ' - Height(kpc) =', item, ' - Index =', index + 1, ' - Weighting = ', AccretedWeighting[GCIdentification])
        MixedHeightGCs.append(item)
        MixedHeightGCIndexes.append(index)
    else:
        AccretedWeighting[GCIdentification] += 4
        print('Name =', GCIdentification, ' - Category = High', ' - Height(kpc) =', item, ' - Index =', index + 1, ' - Weighting = ', AccretedWeighting[GCIdentification])
        HighGCHeights.append(item)
        HighGCIndexes.append(index)
```

```python
# METALLICITY
MergedDataArray = np.array(MergedData, dtype=object)
MetalRichGCS = [] #list of 0s to add metallicities of low metallicity GCs after they are identified by the loop
MetalRichGCIndexes = [] #same as above but to add the indexes 
MixedMetallicityGCs = []
MixedMetallicityGCIndexes = []
MetalPoorGCs = []
MetalPoorGCIndexes = [] #to add other GCs to to help with plotting

GCMetallicity = MergedDataArray[:,29]

for index, item in enumerate(GCMetallicity):
    GCIdentification = MergedDataArray[index,0]
    if -2.5 < item < -1.01:
        AccretedWeighting[GCIdentification] += 5
        print('Name =', GCIdentification, ' - Category = Metal Poor', ' - Metallicity =', item, ' - Index =', index + 1, ' - Weighting = ', AccretedWeighting[GCIdentification])
        MetalPoorGCs.append(item)
        MetalPoorGCIndexes.append(index)
    elif item > -1.0:
        AccretedWeighting[GCIdentification] += 2
        print('Name =', GCIdentification, ' - Category = Metal Rich', ' - Metallicity =', item, ' - Index =', index + 1, ' - Weighting = ', AccretedWeighting[GCIdentification])
        MetalRichGCS.append(item)
        MetalRichGCIndexes.append(index)
```

```python
# LOCAL STANDARD OF REST VELOCITY
MergedDataArray = np.array(MergedData, dtype=object)
QuickGCs = [] #list of 0s to add metallicities of low metallicity GCs after they are identified by the loop
QuickGCIndex = [] #same as above but to add the indexes 
MixedSpeedGCs = []
MixedSpeedGCIndex = []
SlowGCs = []
SlowGCIndex = [] #to add other GCs to to help with plotting

GCLSRVelocity = MergedDataArray[:,13]

for index, item in enumerate(GCLSRVelocity):
    GCIdentification = MergedDataArray[index,0]
    if np.abs(item) < 100:
        AccretedWeighting[GCIdentification] += 0
        print('Name =', GCIdentification, ' - Category = Slow', ' - LSR Velocity =', item, ' - Index =', index + 1, ' - Weighting = ', AccretedWeighting[GCIdentification])
        SlowGCs.append(item)
        SlowGCIndex.append(index)
    elif np.abs(item) > 200:
        AccretedWeighting[GCIdentification] += 2
        print('Name =', GCIdentification, ' - Category = Fast', ' - LSR Velocity =', item, ' - Index =', index + 1, ' - Weighting = ', AccretedWeighting[GCIdentification])
        QuickGCs.append(item)
        QuickGCIndex.append(index)
    else:
        AccretedWeighting[GCIdentification] += 1
        print('Name =', GCIdentification, ' - Category = Mixed Speed', ' - LSR Velocity =', item, ' - Index =', index + 1, ' - Weighting = ', AccretedWeighting[GCIdentification])
        MixedSpeedGCs.append(item)
        MixedSpeedGCIndex.append(index)
```
