```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
HarrisPartI = pd.read_csv('HarrisPartI.csv') # Stores the data of the csv file in the variable HarrisPartI
RightAscension = HarrisPartI['RA'] # Stores the column specfic data under each variable based on the columns name in the csv file using ['column name']
Declination = HarrisPartI['DEC']
GalacticLongitude = HarrisPartI['L']
GalacticLatitude = HarrisPartI['B']
HeliocentricDistance = HarrisPartI['R_Sun']
GalactocentricDistance = HarrisPartI['R_gc']

print(HarrisPartI)
```

This file includes locational data for each globular cluster in terms of right ascension and declination, galactic latitude and longitude, heliocentric distance, galactocentric distance, and galactic distance components including X pointing in the direction of the Galactic centre, Y pointing in the direction of Galactic rotation, and Z pointing towards the north galactic pole.

```python
plt.scatter(GalacticLongitude, GalacticLatitude)   # Makes a scatter plot with Galactic Longitude on the X axis and Galactic Latitutde on the Y axis
plt.xlabel("Galactic Longitude") 
plt.ylabel("Galactic Latitude")
plt.title("Galactic Longitude (L) vs Galactic Latitude (B)") 
plt.show()
```

Galactic centre seen here at (0,0) with a much higher density than the rest of the graph around that point, outliers seen far above and below the galactic plane below -75 and above 75 galactic longitude

```python
plt.scatter(HeliocentricDistance,GalacticLongitude)
plt.xlabel("Heliocentric Distance")
plt.ylabel("Galactic Longitude")
plt.title("Heliocentric Distance(R_sun) vs Galactic Longitude (L)")
plt.show()

```

Several Heliocentric Distance outliers, most clusters close to the sun collect near the galactic centre at smaller galactic longitude differences from the centre. Several globular clusters much farther away from the sun in longitudes behind the sun (around 180)

```python

plt.scatter(HeliocentricDistance,GalacticLatitude)
plt.xlabel("Heliocentric Distance")
plt.ylabel("Galactic Latitude")
plt.title("Heliocentric Distance(R_sun) vs Galactic Latitude (B)")
plt.show()
```

Globular clusters far above or below the galactic plane (galactic latitude=0) seem to be found at all heliocentric distances, globular clusters very close to the sun can be found at above 75 and below -75 galactic longitude

```python

plt.scatter(GalactocentricDistance,GalacticLongitude)
plt.xlabel("Galactocentric Distance")
plt.ylabel("Galactic Longitude")
plt.title("Galactocentric Distance(R_gc) vs Galactic Longitude (L)")
plt.show()


plt.scatter(GalactocentricDistance,GalacticLatitude)
plt.xlabel("Galactocentric Distance")
plt.ylabel("Galactic Latitude")
plt.title("Galactocentric Distance(R_gc) vs Galactic Latitude (B)")
plt.show()
```

Same trends seen using galactocentric distance instead of heliocentric, Galactocentric distance vs Heliocentric distance doesn't show anything relevant, linear. These graphs are really just indicating where the globular clusters are in the galaxy, this data will be most useful when we consider the properties of the globular clusters in conjunction with their position in the galaxy to see their relationships.
