```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

Importing Data from CSV file

```python

HarrisPartI = pd.read_csv('HarrisPartI.csv') # assigns the csv data to variable HarrisPartI
RightAscension = HarrisPartI['RA'] # assigns column specific data to variables
Declination = HarrisPartI['DEC']
GalacticLongitude = HarrisPartI['L']
GalacticLatitude = HarrisPartI['B']
HeliocentricDistance = HarrisPartI['R_Sun']
GalactocentricDistance = HarrisPartI['R_gc']

print(HarrisPartI)
```

Below is a collection of graphs of different variables in the data set to get an idea of how they're related

```python

plt.scatter(GalacticLongitude, GalacticLatitude) # plot with x = galactic longitude, y = galactic latitude
plt.xlabel("Galactic Longitude")
plt.ylabel("Galactic Latitude")
plt.title("Galactic Longitude (L) vs Galactic Latitude (B)")
plt.show()
```

This plot is essentially a map of the globular clusters throughout the galaxy, Galactic centre at (0,0), clearly higher density of globular clusters the closer you get to the galactic centre, clusters far above and below the plane of the galaxy with clusters above 75 and below -75 galactic latitude, clusters at all angles, galactic longitude 180.

```python
plt.scatter(HeliocentricDistance,GalacticLongitude)
plt.xlabel("Heliocentric Distance")
plt.ylabel("Galactic Longitude")
plt.title("Heliocentric Distance(R_sun) vs Galactic Longitude (L)")
plt.show()
```

clusters naturally focused at a closer heliocentric distance because its closer to the galactic centre, same trend obsered if using galactocentric distance instead of heliocentric (see galactocentric distance vs galactic longitude graph below), several clusters at a very high heliocentric distance, focused around the galactic longitudes directly opposite the galactic centre ~180 galactic longitude

```python
plt.scatter(HeliocentricDistance,GalacticLatitude)
plt.xlabel("Heliocentric Distance")
plt.ylabel("Galactic Latitude")
plt.title("Heliocentric Distance(R_sun) vs Galactic Latitude (B)")
plt.show()
```

Globular clusters can be seen at high and low galactic latitudes at varying heliocentric distances with globular clusters close to the galactic centre seen above 75 and below -75 galactic latitude, clusters around the galactic centre focus more around galactic latitude 0 with that trend less aparent the farther the heliocentric distance, more scattering the farther out you go, yet still high range of latitudes at galactic centre. This trend is to be expected, milky way is a disk galaxy so more mass focues on the galactic plane specifically around the core regions, pulling globular clusters in towards galactic latitude 0, but further out, despite it being more massive than the inner regions of the galaxy due to majority of the mass of the galaxy being the dark matter halo, the dark matter halo is spherical so it doesnt have the latitude mass bias and hence the globular clusters are much more scattered.

```python
plt.scatter(GalactocentricDistance,GalacticLongitude)
plt.xlabel("Galactocentric Distance")
plt.ylabel("Galactic Longitude")
plt.title("Galactocentric Distance(R_gc) vs Galactic Longitude (L)")
plt.show()
```

same trends as before just Galactocentric instead of heliocentric

```python
plt.scatter(GalactocentricDistance,GalacticLatitude)
plt.xlabel("Galactocentric Distance")
plt.ylabel("Galactic Latitude")
plt.title("Galactocentric Distance(R_gc) vs Galactic Latitude (B)")
plt.show()
```

again same trends as heliocentric counterpart
